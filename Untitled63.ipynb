```python
con.execute('''
CREATE OR REPLACE VIEW origination_filtered AS

WITH base AS (
    SELECT
        loan_id,

        -- CREDIT SCORE
        CASE 
            WHEN TRY_CAST(credit_score AS INTEGER) BETWEEN 300 AND 850 
            THEN TRY_CAST(credit_score AS INTEGER)
            ELSE NULL 
        END AS credit_score,

        -- DATES
        STRPTIME(first_payment_date || '01', '%Y%m%d') AS loan_start_date,
        STRPTIME(maturity_date || '01', '%Y%m%d') AS maturity_date,

        -- FLAGS
        CASE 
            WHEN first_time_homebuyer_flag IN ('Y','N') THEN first_time_homebuyer_flag
            ELSE NULL 
        END AS first_time_homebuyer_flag,

        -- RAW VALUES NEEDED LATER
        ltv,
        cltv,

        -- OTHER FIELDS (unchanged)
        TRY_CAST(msa AS INTEGER) AS msa,

        CASE 
            WHEN TRY_CAST(mi_percentage AS DOUBLE) = 0 THEN 0
            WHEN TRY_CAST(mi_percentage AS DOUBLE) BETWEEN 1 AND 55 THEN TRY_CAST(mi_percentage AS DOUBLE)
            ELSE NULL
        END AS mi_percentage,

        CASE 
            WHEN TRY_CAST(num_units AS INTEGER) BETWEEN 1 AND 4 THEN TRY_CAST(num_units AS INTEGER)
            ELSE NULL
        END AS num_units,

        CASE 
            WHEN occupancy_status IN ('P','I','S') THEN occupancy_status
            ELSE NULL
        END AS occupancy_status,

        TRY_CAST(dti AS DOUBLE) AS dti_raw,

        TRY_CAST(original_upb AS DOUBLE) AS original_upb,
        TRY_CAST(original_interest_rate AS DOUBLE) AS original_interest_rate,

        CASE 
            WHEN channel IN ('R','B','C','T') THEN channel
            ELSE NULL 
        END AS channel,

        CASE WHEN ppm_flag IN ('Y','N') THEN ppm_flag ELSE NULL END AS ppm_flag,
        CASE WHEN amortization_type IN ('FRM','ARM') THEN amortization_type ELSE NULL END AS amortization_type,

        property_state,

        CASE 
            WHEN property_type IN ('CO','PU','MH','SF','CP') THEN property_type
            ELSE NULL 
        END AS property_type,

        TRY_CAST(postal_code AS INTEGER) AS postal_code,

        CASE 
            WHEN loan_purpose IN ('P','C','N','R') THEN loan_purpose
            ELSE NULL 
        END AS loan_purpose,

        TRY_CAST(original_loan_term AS INTEGER) AS original_loan_term,

        CASE 
            WHEN TRY_CAST(num_borrowers AS INTEGER) BETWEEN 1 AND 10 
            THEN TRY_CAST(num_borrowers AS INTEGER)
            ELSE NULL 
        END AS num_borrowers,

        seller_name,
        servicer_name,

        CASE 
    WHEN super_conforming_flag = 'Y' THEN 1
    WHEN super_conforming_flag = 'N' THEN 0
    ELSE NULL
END AS super_conforming_flag,
        pre_refinance_loan_id,

        CASE 
            WHEN program_type IN ('H','F','R') THEN program_type
            ELSE NULL 
        END AS program_type,

        CASE WHEN relief_refinance_flag = 'Y' THEN 1 ELSE 0 END AS relief_refinance_flag,

        CASE 
            WHEN TRY_CAST(valuation_method AS INTEGER) IN (1,2,3,4) 
            THEN TRY_CAST(valuation_method AS INTEGER)
            ELSE NULL 
        END AS valuation_method,

        CASE WHEN interest_only_flag IN ('Y','N') THEN interest_only_flag ELSE NULL END AS interest_only_flag,

        CASE 
            WHEN mi_cancellation_flag IN ('Y','N') THEN mi_cancellation_flag
            ELSE NULL 
        END AS mi_cancellation_flag

    FROM origination_clean
),

ltv_stage AS (
    SELECT *,
        CASE 
            WHEN loan_start_date < DATE '2018-04-01'
                 AND TRY_CAST(ltv AS DOUBLE) BETWEEN 6 AND 105
            THEN TRY_CAST(ltv AS DOUBLE)

            WHEN loan_start_date >= DATE '2018-04-01'
                 AND TRY_CAST(ltv AS DOUBLE) BETWEEN 1 AND 998
            THEN TRY_CAST(ltv AS DOUBLE)

            ELSE NULL
        END AS ltv_clean
    FROM base
)

SELECT 
    loan_id,
    credit_score,
    loan_start_date,
    maturity_date,
    first_time_homebuyer_flag,
    msa,
    mi_percentage,
    num_units,
    occupancy_status,
    original_upb,
    original_interest_rate,
    channel,
    ppm_flag,
    amortization_type,
    property_state,
    property_type,
    postal_code,
    loan_purpose,
    original_loan_term,
    num_borrowers,
    seller_name,
    servicer_name,
    super_conforming_flag,
    pre_refinance_loan_id,
    program_type,
    relief_refinance_flag,
    valuation_method,
    interest_only_flag,
    mi_cancellation_flag,
    ltv_clean,

    -- FINAL CLTV USING CLEAN LTV
    CASE 
        WHEN loan_start_date < DATE '2018-04-01'
             AND TRY_CAST(cltv AS DOUBLE) BETWEEN 6 AND 200
             AND ltv_clean IS NOT NULL
             AND TRY_CAST(cltv AS DOUBLE) >= ltv_clean
        THEN TRY_CAST(cltv AS DOUBLE)

        WHEN loan_start_date >= DATE '2018-04-01'
             AND TRY_CAST(cltv AS DOUBLE) BETWEEN 1 AND 998
             AND ltv_clean IS NOT NULL
             AND TRY_CAST(cltv AS DOUBLE) >= ltv_clean
        THEN TRY_CAST(cltv AS DOUBLE)

        ELSE NULL
    END AS cltv_clean,

    -- CLEAN DTI (we'll refine next step)
    CASE 
    -- HARP loans → always NULL
    WHEN relief_refinance_flag = 1 THEN NULL

    -- Valid DTI range
    WHEN dti_raw BETWEEN 1 AND 65 THEN dti_raw

    ELSE NULL
END AS dti

FROM ltv_stage;
''')
```

```python
con.execute("""
CREATE OR REPLACE VIEW performance_clean AS
SELECT 
    column00  AS loan_id,

    -- Dates
    CAST(column01 AS INTEGER) AS reporting_period,

    -- Balances
    CAST(column02 AS DOUBLE) AS current_actual_upb,

    -- Target / behavior
    column03 AS delinquency_status,

    -- Time features
    CAST(column04 AS INTEGER) AS loan_age,
    CAST(column05 AS INTEGER) AS remaining_months_to_maturity,

    -- Events
    CAST(column06 AS INTEGER) AS defect_settlement_date,
    column07 AS modification_flag,
    column08 AS zero_balance_code,
    CAST(column09 AS INTEGER) AS zero_balance_effective_date,

    -- Rates
    CAST(column10 AS DOUBLE) AS current_interest_rate,

    -- UPB breakdown
    CAST(column11 AS DOUBLE) AS non_interest_bearing_upb,

    -- Payment tracking
    CAST(column12 AS INTEGER) AS ddlpi,

    -- Recoveries / Loss
    CAST(column13 AS DOUBLE) AS mi_recoveries,
    column14 AS net_sale_proceeds,
    CAST(column15 AS DOUBLE) AS non_mi_recoveries,

    -- Expenses
    CAST(column16 AS DOUBLE) AS total_expenses,
    CAST(column17 AS DOUBLE) AS legal_costs,
    CAST(column18 AS DOUBLE) AS maintenance_costs,
    CAST(column19 AS DOUBLE) AS taxes_insurance,
    CAST(column20 AS DOUBLE) AS misc_expenses,

    -- Loss metric
    CAST(column21 AS DOUBLE) AS actual_loss,

    -- Modification economics
    CAST(column22 AS DOUBLE) AS cumulative_modification_cost,
    column23 AS step_modification_flag,
    column24 AS payment_deferral_flag,

    -- Risk metrics
    CAST(column25 AS DOUBLE) AS estimated_ltv,

    -- Default mechanics
    CAST(column26 AS DOUBLE) AS zero_balance_removal_upb,
    CAST(column27 AS DOUBLE) AS delinquent_accrued_interest,

    -- Flags
    column28 AS disaster_flag,
    column29 AS borrower_assistance_status,

    -- Monthly mod cost
    CAST(column30 AS DOUBLE) AS current_month_mod_cost,

    -- Core balance
    CAST(column31 AS DOUBLE) AS interest_bearing_upb

FROM 'C:/Users/snarb/OneDrive/Desktop/loan_pipeline/parquet/performance/**/*.parquet';
""")
```

```python
con.execute('''
CREATE OR REPLACE VIEW performance_filtered AS
SELECT 
    loan_id,
    reporting_period,
    current_actual_upb,
    
    CASE 
        WHEN delinquency_status = 'RA' THEN 999  -- REO bucket
        WHEN TRY_CAST(delinquency_status AS INTEGER) IS NOT NULL 
            THEN CAST(delinquency_status AS INTEGER)
        ELSE NULL
    END AS delinquency_status_num,

    CASE 
        WHEN delinquency_status = 'RA' THEN 'REO'
        WHEN TRY_CAST(delinquency_status AS INTEGER) >= 3 THEN 'DEFAULT'
        WHEN TRY_CAST(delinquency_status AS INTEGER) IN (1,2) THEN 'DELINQUENT'
        WHEN delinquency_status = '0' THEN 'CURRENT'
        ELSE 'UNKNOWN'
    END AS delinquency_bucket,

    -- TARGET VARIABLES
   

    CASE 
        WHEN delinquency_status = 'RA' THEN 1
        WHEN TRY_CAST(delinquency_status AS INTEGER) >= 3 THEN 1
        ELSE 0
    END AS default_flag,

    CASE 
        WHEN zero_balance_code IS NOT NULL THEN 1
        ELSE 0 
    END AS terminated_flag,

    -- =========================
    -- LOAN STATE (THIS IS GOLD)
    -- =========================
    CASE 
        WHEN zero_balance_code IS NOT NULL THEN 'TERMINATED'
        WHEN delinquency_status = 'RA' THEN 'REO'
        WHEN TRY_CAST(delinquency_status AS INTEGER) >= 3 THEN 'DEFAULT'
        WHEN TRY_CAST(delinquency_status AS INTEGER) > 0 THEN 'DELINQUENT'
        ELSE 'CURRENT'
    END AS loan_state,

    -- =========================
    -- TIME FEATURES
    -- =========================
    loan_age,
    remaining_months_to_maturity,

    -- =========================
    -- RATES
    -- =========================
    current_interest_rate,

    -- =========================
    -- LTV CLEANING
    -- =========================
    CASE 
        WHEN estimated_ltv BETWEEN 1 AND 998 THEN estimated_ltv
        ELSE NULL
    END AS estimated_ltv,

    -- =========================
    -- IMPORTANT EDGE FLAGS (FROM VALIDATION)
    -- =========================

    -- RA without termination (keep track)
    CASE 
        WHEN delinquency_status = 'RA' 
             AND zero_balance_code IS NULL THEN 1
        ELSE 0
    END AS reo_no_termination_flag,

    -- UPB inconsistency
    CASE 
        WHEN current_actual_upb = 0 
             AND zero_balance_code IS NULL THEN 1
        ELSE 0
    END AS zero_upb_active_flag,

    -- Extreme delinquency
    CASE 
        WHEN TRY_CAST(delinquency_status AS INTEGER) > 60 THEN 1
        ELSE 0
    END AS extreme_dq_flag,
    CASE 
    WHEN zero_balance_code = '01' THEN 1
    ELSE 0
END AS prepaid_flag,

    -- =========================
    -- RAW BACKUP (ALWAYS KEEP)
    -- =========================
    delinquency_status,
    zero_balance_code

FROM performance_clean;
''')
```

```python
############################Create Forward PD (12-month horizon):

############################For each loan at time t, check:
############################Does it default in next 12 months?
```

```python
con.execute('''
CREATE OR REPLACE VIEW forward_pd_12m AS

SELECT 
    p1.loan_id,
    p1.reporting_period,

    -- =========================
    -- TARGET: DEFAULT IN NEXT 12M
    -- =========================
    MAX(
        CASE 
            WHEN p2.default_flag = 1 THEN 1
            ELSE 0
        END
    ) AS default_in_12m,

    -- =========================
    -- OPTIONAL: PREPAY IN NEXT 12M
    -- =========================
    MAX(
        CASE 
            WHEN p2.prepaid_flag = 1 THEN 1
            ELSE 0
        END
    ) AS prepaid_in_12m

FROM performance_filtered p1

LEFT JOIN performance_filtered p2
ON p1.loan_id = p2.loan_id
AND p2.reporting_period > p1.reporting_period
AND p2.reporting_period <= p1.reporting_period + 12

GROUP BY 
    p1.loan_id,
    p1.reporting_period;
''')
```

```python
con.execute('''
CREATE OR REPLACE VIEW forward_pd_12m_filtered AS

SELECT f.*
FROM forward_pd_12m f
JOIN performance_filtered p
ON f.loan_id = p.loan_id
AND f.reporting_period = p.reporting_period

WHERE p.loan_state = 'CURRENT';
''')
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


# =========================
# KS DATA FUNCTION
# =========================

def ks_curve(y_true, y_pred):

    data = pd.DataFrame({
        'y': y_true,
        'p': y_pred
    })

    # Sort descending by probability
    data = data.sort_values('p', ascending=False).reset_index(drop=True)

    # Cumulative bad and good
    data['cum_bad'] = data['y'].cumsum() / data['y'].sum()

    data['cum_good'] = (
        (1 - data['y']).cumsum()
        / (1 - data['y']).sum()
    )

    # KS distance
    data['ks'] = data['cum_bad'] - data['cum_good']

    # Population %
    data['population'] = np.arange(1, len(data)+1) / len(data)

    return data


# =====================================================
# CREATE KS DATA
# =====================================================

ks_val_data = ks_curve(y_val, pred_val)

ks_fwd_data = ks_curve(
    test_fwd['default_in_12m'],
    pred_test_fwd
)

ks_back_data = ks_curve(
    test_back['default_in_12m'],
    pred_test_back
)


# =====================================================
# COMBINED KS PLOT
# =====================================================

plt.figure(figsize=(9,6))

plt.plot(
    ks_val_data['population'],
    ks_val_data['ks'],
    label=f'Validation KS = {ks_val:.4f}'
)

plt.plot(
    ks_fwd_data['population'],
    ks_fwd_data['ks'],
    label=f'Forward Test KS = {ks_fwd:.4f}'
)

plt.plot(
    ks_back_data['population'],
    ks_back_data['ks'],
    label=f'Back Test KS = {ks_back:.4f}'
)

plt.axhline(0, linestyle='--')

plt.title('KS Statistic Comparison')
plt.xlabel('Population %')
plt.ylabel('KS Value')

plt.legend()
plt.grid(True)

plt.show()
```

```python
def ks_stat(y_true, y_pred):
    df = pd.DataFrame({'y': y_true, 'p': y_pred})
    df = df.sort_values('p')
    
    df['bad'] = (df['y'] == 1).astype(int)
    df['good'] = (df['y'] == 0).astype(int)
    
    df['cum_bad'] = df['bad'].cumsum() / df['bad'].sum()
    df['cum_good'] = df['good'].cumsum() / df['good'].sum()
    
    return np.max(np.abs(df['cum_bad'] - df['cum_good']))
```
